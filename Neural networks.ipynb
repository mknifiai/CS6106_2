### Universal approximation theorem

- The number of hidden units in a shallow network is a measure of **the network capacity**.
- With ReLU activation functions, the output of a network with D hidden units has at most D joints and so is a piecewise linear function with at most D + 1 linear regions.
- As we add more hidden units, the model can approximate more complex functions.

## 3. Convolutional networks

**Images properties** 

- They are high-dimensional.

    - A typical image for a classification task contains 224×224 RGB values (i.e., 150,528 input dimensions)
      
    - With fully connected networks, this poses obvious practical problems in terms of the required training data, memory, and computation. 

- Local pattern: nearby image pixels are statistically related.

    - spatial relations between local patterns are informative
    
    - fully connected networks have no notion of “nearby” and treat the relationship between every input equally


- The interpretation of an image is stable under geometric transformations.

### Invariance and equivariance

**Invariance**

- A function $f[x]$ of an image $x$ is invariant to a transformation $t[x]$ if: 

 $\qquad \qquad f[t[x]] = f[x].$
     
     - Networks for image classification should be invariant to geometric transformations of the image 

**equivariance**

- A function $f[x]$ of an image $x$ is equivariant or covariant to a transformation $t[x]$ if:

$ \qquad \qquad f[t[x]] = t[f[x]]$.

    - Networks for per-pixel image segmentation should be equivariant to transformations 

<img src="img/fig5.png" width=500 height=500 />

**Convolutional layers** (supposed to be equivariant to transformations)

    - process each local image region independently, using parameters shared across the whole image. 

    - They use fewer parameters than fully connected layers, exploit the spatial relationships between nearby pixels, and don’t have to re-learn the interpretation of the pixels at every position.

**pooling layers** are designed to provide invariance

## 3.1. Convolutional networks for 1D inputs

- In 1D, a convolution transforms an input vector $x$ into an output vector $z$ so that each output $z_{i}$ is a weighted sum of nearby inputs.
- The same weights are used at every position and are collectively called the convolution kernel or filter.
- The size of the region over which inputs are combined is termed the kernel size.
- Can be applied to financial time series, audio, and text

**For a kernel size of three, we have:**

$z_{i} = ω_{1}x_{i−1} + ω_{2}x_{i} + ω_{3}x_{i+1}$

where $ω = [ω_{1},ω_{2},ω_{3}]^{T}$ is the kernel

### 3.1.1 Padding

Two approaches:
- The first is to pad the edges of the inputs with new values and proceed as usual. 
    - Zero-padding assumes the input is zero outside its valid range (figure 10.2c)


- The second approach is to discard the output positions where the kernel exceeds the range of input positions
    - Disadvantage: the representation decreases in size

### 3.1.2 Stride, kernel size, and dilation

- When we evaluate the output at every position, we term this **a stride** of one.
- If we have a stride of two, we create roughly half the number of outputs (figure 10.3a–b)

- **The kernel size** can be increased to integrate over a larger area (figure 10.3c)
   - Increasing the kernel size has the disadvantage of requiring more weights

- **Dilated or atrous convolutions**: the kernel values are interspersed with zeros.
    - For example, we can turn a kernel of size five into a dilated kernel of size three by setting the second and fourth elements to zero (figure 10.3d)
    - **the dilation rate**: the number of zeros we intersperse between the weights determines.

### 3.1.4 Channels

- If we only apply a single convolution, information will likely be lost
- Hence, it is usual to compute several convolutions in parallel.
    - Each convolution produces a new set of hidden variables, termed a feature map or channel.

- and be applied to the three input channels at each of the 3×3 positions to create a 2D output that is the same height and width as the input image (assuming zero-padding)

- If the kernel size is $K × K$ , and there are $C_{i}$ input channels, each output channel is a weighted sum of $C_{i} × K × K$ quantities plus one bias.

- It follows that to compute $C_{o}$ output channels, we need $C_{i} × C_{o} × K × K$ weights and $C_{o}$ biases.

## 5. Downsampling and upsampling

### 5.1 Downsampling

- There are three approaches to scaling down both dimensions by a factor of two:

1. use a stride of two
2. max pooling (induces some invariance to translation)
3. mean pooling or average pooling

- For all approaches, we apply downsampling separately to each channel,
- so the output has half the width and height but the same number of channels.


<img src="img/fig15.png" width=500 height=500 />

### 5.2 Upsampling


1. duplicate all the channels at each spatial position four times
2. max unpooling (is used where we have previously used a max pooling operation for downsampling)
3. uses bilinear interpolation to fill in the missing values between the points where we have samples

4. use transposed convolution (called fractionally-strided convolution or deconvolution)
